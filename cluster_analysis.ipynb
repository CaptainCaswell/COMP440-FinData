```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 100)
pd.set_option("display.float_format", "{:.4f}".format)

FILE = "data/clustered_data_gen2_depth4.parquet"

# Load final depth
df = pd.read_parquet( FILE )

print(df.shape)
df.head()
```

```python
summary = pd.read_csv(
    "data/cluster_summary_depth1.csv"
)

summary
```

```python
BEST_CLUSTER = 0

best = df[df["cluster"] == BEST_CLUSTER].copy()
rest = df[df["cluster"] != BEST_CLUSTER].copy()

print("Best cluster:")
print(best.shape)

print("\nRest:")
print(rest.shape)
```

```python
best_returns = best["future_5y_return"]
rest_returns = rest["future_5y_return"]

comparison = pd.DataFrame({
    "Best Cluster": best_returns.describe(),
    "Rest": rest_returns.describe()
})

comparison
```

```python
# Select numeric features
numeric_cols = df.select_dtypes(include=np.number).columns

TARGET_COL = "future_5y_return"
ID_COLS = ["ticker", "date", "sector"]

PRIM_EXCLUDE = {
    "price",
    "sector_size",
    "rows",
    "sector_size_market",
    "rows_market"
}

SEC_EXCLUDE = {
    "excess_ret_1y", "excess_ret_5y",
    "trend_vs_sector_1y", "trend_vs_sector_5y",
    "drawdown_rel_1y", "drawdown_rel_5y",
    "excess_vs_market_1y", "trend_vs_market_1y",
    "risk_adjusted_1y", "risk_adjusted_5y",
    "quality_score",
    "sector_Unknown",
    "pe",
    "cluster"
}

exclude = set( ID_COLS ) | { TARGET_COL } | PRIM_EXCLUDE | SEC_EXCLUDE

features = [c for c in numeric_cols if c not in exclude and not c.endswith( "_market" )]


# Compare averages
feature_comparison = pd.DataFrame({
    "best_mean": best[features].mean(),
    "rest_mean": rest[features].mean()
})


feature_comparison["difference"] = (
    feature_comparison["best_mean"]
    -
    feature_comparison["rest_mean"]
)


feature_comparison["effect_size"] = (
    feature_comparison["difference"]
    /
    df[features].std()
)


feature_comparison.sort_values(
    "effect_size",
    key=lambda x: x.abs(),
    ascending=False
).head(30)
```
