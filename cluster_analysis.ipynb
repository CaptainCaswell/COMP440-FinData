```python
import numpy as np
import pandas as pd

# ==========================================
# Load depth 1 clustering results
# ==========================================

df = pd.read_parquet(
    "data/clustered_data_depth1.parquet"
)

print("Dataset loaded:")
print("Rows:", len(df))
print("Tickers:", df["ticker"].nunique())

print("\nCluster counts:")
print(df["cluster"].value_counts())


# ==========================================
# Split clusters
# ==========================================

BEST_CLUSTER = 1

best = df[df["cluster"] == BEST_CLUSTER]
rest = df[df["cluster"] != BEST_CLUSTER]

print("\nBest cluster rows:", len(best))
print("Rest rows:", len(rest))


# ==========================================
# Select numeric features
# ==========================================

numeric_cols = df.select_dtypes(include=np.number).columns

exclude = [
    "cluster",
    "future_5y_return"
]

features = [
    c for c in numeric_cols
    if c not in exclude
]

print("\nFeatures analyzed:", len(features))


# ==========================================
# Compare feature means
# ==========================================

comparison = pd.DataFrame()

comparison["best_mean"] = best[features].mean()
comparison["rest_mean"] = rest[features].mean()

comparison["difference"] = (
    comparison["best_mean"]
    -
    comparison["rest_mean"]
)

# Standardized difference
comparison["effect_size"] = (
    comparison["difference"]
    /
    df[features].std()
)

comparison = comparison.sort_values(
    "effect_size",
    key=lambda x: x.abs(),
    ascending=False
)


# ==========================================
# Display results
# ==========================================

comparison.head(30)
```
