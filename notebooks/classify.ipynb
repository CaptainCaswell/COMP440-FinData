```python
# Load cluster
import pandas as pd
import numpy as np

df = pd.read_parquet( "../logs/classification/test_scored_gen1.parquet" )

print( "(Rows, Columns)" )
print( df.shape )
```

```python
rf_top = df.nlargest(50, "rf_probability")

rf_top[
    [
        "ticker",
        "date",
        "rf_probability",
        "future_excess_5y",
        "beats_market_5y"
    ]
]
```

```python
rf_top10 = (
    df.sort_values("rf_probability", ascending=False)
      .head(int(len(df)*0.10))
)

rf_top10["ticker"].value_counts().head(50)
```

```python
stock_summary = (
    rf_top10.groupby("ticker")
    .agg(
        times_selected=("ticker", "count"),
        avg_probability=("rf_probability", "mean"),
        median_probability=("rf_probability", "median"),
        avg_future_return=("future_excess_5y", "mean"),
        median_future_return=("future_excess_5y", "median"),
        beat_rate=("beats_market_5y", "mean"),
    )
)

stock_summary.sort_values(
    ["beat_rate", "avg_future_return"],
    ascending=False
).head(50)
```
