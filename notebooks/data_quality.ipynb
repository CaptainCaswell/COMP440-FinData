```python
import pandas as pd

df = pd.read_parquet("../data/data.parquet")

beat_spy = (df["future_excess_5y"] > 0).mean()

print( f" Percentage of observations that beat SPY: {beat_spy:.2%}" )
```

```python
print( "Distribution of excess")

df["future_excess_5y"].quantile(
    [0.01,0.05,0.10,0.25,0.50,0.75,0.90,0.95,0.99]
)
```

```python
print( "Confirm excess returns")
check = df[df["ticker"].isin(["SPY", "AAPL", "MSFT"])]

check[[
    "ticker",
    "date",
    "future_ret_5y",
    "future_excess_5y"
]].groupby("ticker").head(5)
```
