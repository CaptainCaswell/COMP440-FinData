```python
import yfinance as yf

symbol = "AAPL"

ticker = yf.Ticker( symbol ) # Try "AAPL" as a stock comparison
info = ticker.info

# Check if the type is 'ETF'
type = info.get('quoteType')

is_etf = type == 'ETF'

print( type )

print(f"Is {ticker} an ETF? {is_etf}") 
```

```python
import json

# Load the files
with open("../data/company_tickers.json", "r") as f:
    data1 = json.load(f)

with open("../data/etf.json", "r") as f:
    data2 = json.load(f)

# Extract tickers from each file
tickers1 = {record["ticker"] for record in data1.values()}
tickers2 = set(data2)

# Compare
only_in_file1 = tickers1 - tickers2
only_in_file2 = tickers2 - tickers1
common = tickers1 & tickers2

print(f"File 1 tickers: {len(tickers1)}")
print(f"File 2 tickers: {len(tickers2)}")
print(f"Common tickers: {len(common)}")
print(f"Only in File 1: {len(only_in_file1)}")
print(f"Only in File 2: {len(only_in_file2)}")
```
