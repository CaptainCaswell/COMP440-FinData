```python
import pandas as pd

info_df = pd.read_parquet("data/raw_info.parquet")

print(f"Total tickers: {len(info_df)}")
print(f"fetch_success=False: {(~info_df['fetch_success']).sum()}")
print(f"trailing_pe missing (fetch succeeded): {info_df[info_df['fetch_success']]['trailing_pe'].isna().sum()}")
print(f"trailing_pe <= 0: {(info_df['trailing_pe'] <= 0).sum()}")
```
